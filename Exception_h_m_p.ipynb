## Exceptions

#### If things go wrong during the execution of the program(runtime). It generally happens when something unforeseen has happend.
    - Exceptions are raised by python runtime
    - You have to take is on the fly

#### Examples
    - Memory overflow
    - Divide by 0 -> logical erro
    - Databse error

# Why is it important to handle exceptions
# how to handle exceptions
# -> Try except block

```python
# let's create a file
with open('sample1.txt','w') as f:
    f.write('Hello World')
```

```python
# try catch demo 
try:
    with open('sample2.txt','r') as f:
        print(f.read())
except:
    print("Sorry file not found : ")
```

```python
# catching specific exception
try:
    m=5
    f = open('sample1.txt','r')
    print(f.read())
    print(m)
    print(5/2)
    L=[1,2,3,4]
    L[100]
except FileNotFoundError:
    print("File not found : ")
except NameError:
    print("Variable not defined : ")
except ZeroDivisionError:
    print("Can't divide by 0")
except Exception as e:
    print(e)
```

```python
# else
try:
    f = open('sample2.txt','r')
except FileNotFoundError:
    print('file not founnd error')
except Exception:
    print("kuchh to lafda hai ")
else:
    print(f.read())
```

```python
# finally
# else
try:
    f = open('sample2.txt','r')
except FileNotFoundError:
    print('file not founnd error')
except Exception:
    print("kuchh to lafda hai ")
else:
    print(f.read())
finally:
    print("Ye to print hoga hi : ")
```

```python
# raise Exception 
# In python programming, exceptions are raised when error occure at runtime.
# We can also manually raise exceptions using the raise kayword.
# We can optionally pass value to the exception to clarity why that exception was raised.
```

```python
class Bank:

  def __init__(self,balance):
    self.balance = balance

  def withdraw(self,amount):
    if amount < 0:
      raise Exception('amount cannot be -ve')
    if self.balance < amount:
      raise Exception('paise nai hai tere paas')
    self.balance = self.balance - amount

obj = Bank(10000)
try:
  obj.withdraw(15000)
except Exception as e:
  print(e)
else:
  print(obj.balance)
```

```python
class MyException(Exception):
    def __init__(self,message):
        print(message)

class Bank:
    def __init__(self,balance):
        self.balance = balance
    
    def withdraw(self,amount):
        if amount <=0:
            raise MyException('Amount cannot be -ve : ')
        if self.balance < amount:
            raise MyException('Paise nhi hai tere paas :')
        self.balance = self.balance - amount
    
obj = Bank(10000)
try:
    obj.withdraw(5000)
except MyException as e:
    pass
else:
    print(obj.balance)
        
```

```python
# simple example
class SecurityError(Exception):

  def __init__(self,message):
    print(message)

  def logout(self):
    print('logout')

class Google:

  def __init__(self,name,email,password,device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self,email,password,device):
    if device != self.device:
      raise SecurityError('bhai teri to lag gayi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('login error')



obj = Google('nitish','nitish@gmail.com','1234','android')

try:
  obj.login('nitish@gmail.com','1234','windows')
except SecurityError as e:
  e.logout()
else:
  print(obj.name)
finally:
  print('database connection closed')

```

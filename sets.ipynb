### Sets
A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:

- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

1. Set mutable hai
- Set ke andar elements add, remove ya update nahi, balki add/remove kiya ja sakta hai.

- s = {1, 2, 3}
- s.add(4)
- s.remove(2)

2. Set ke elements hashable hone chahiye
  - Set apne elements ko hashing ke through store aur locate karta hai. Isliye element ka hash stable hona chahiye.

  - s = {1, "hello", (2, 3)}   # valid

3. Mutable elements set mein nahi ho sakte
  - list, dict, aur set hashable nahi hote, isliye unhe set ke element ke roop mein nahi rakh sakte.

  - s = {[1, 2]}   # TypeError

4. Important interview distinction

  - Set itself is mutable, but its elements must be hashable.

-- Matlab:

  - Set → Mutable
  ↓
  - Elements → Hashable

-- Interview answer:

  - “A set is mutable because we can add or remove elements. However, its individual elements must be hashable because Python uses hashing to efficiently locate elements inside the set.”

#### Set Operation

```python
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1 | s2

# union(|)
# s1.union(s2)
# Intersection(&)
# s1.intersection(s2)
# s1 & s2
# s1-s2:- s1 mein jo elements hain, lekin s2 mein nahi hain.
# Pehle wale set se second wale set ke common elements hatao.
print(s1-s2)
# s2-s1 :- s2 mein jo elements hain, lekin s1 mein nahi hain.
print(s2-s1)
# Dono sets ke aise elements jo sirf ek set mein hain, dono mein common nahi hain.
print(s1 ^ s2)
s1.symmetric_difference(s2)
# Membership Test
# Membership operators in and not in are used to check whether an element exists in a set.
print(1 in s1)
print(1 not in s1)

# Itretions
for i in s1:
    print(i)
    


```

### Set Functions

```python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
len(s)
sum(s)
min(s)
max(s)
sorted(s,reverse=True)
```

```python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2
# New set return karta hai and Original set modify nhi karta hai
s3=s1.union(s1)

# New set return nhi karta hai and Original set modify karta hai
s4=s1.update(s2)
print(s4)
print(s1)
print(s2)
print(s3)
```

union() returns a new set containing elements from both sets without modifying the original sets, whereas update() modifies the original set in place by adding elements from another iterable.

```python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Dono sets mein jo common elements hain, unka new set banao.
s3 = s1.intersection(s2)
print(s3)

# Original sets ko modify karna hai toh intersection_update ka use karein.
s1.intersection_update(s2)
print(s1)
print(s2)
```

intersection() returns a new set containing common elements without modifying the original set, while intersection_update() modifies the original set to keep only common elements.

```python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1.difference(s2)
# s1 mein jo elements hain, lekin s2 mein nahi hain.
s3 = s1.difference(s2)
print(s3)

s1.difference_update(s2)
# s1 mein jo elements hain, lekin s2 mein nahi hain, unko hata do.
print(s1)
print(s2)
```

#####  difference() returns a new set containing elements present in the first set but not in the second, while difference_update() removes those elements from the original set in place.

```python
# symmetric_difference() → new set return karta hai
# symmetric_difference_update() → original set ko modify karta hai
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Dono sets ke unique elements lo, lekin common elements ko hata do.
s3 = s1.symmetric_difference(s2)
print(s3)
s1 ^ s2

s1.symmetric_difference_update(s2)
s1 ^= s2
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}

# Kya dono sets mein ek bhi common element nahi hai?
s1.isdisjoint(s2)
# isdisjoint() returns True if two sets have no elements in common; otherwise, it returns False.
```

```python
# Kya s1 ke saare elements s2 mein present hain?
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issubset(s2)
```

```python

s1 = {1,2}
s2 = {1,2,3,4}
s2.issuperset(s1)
# kya s2 ke andar s1 ke saare elements present hain?
```

Interview Definitions

isdisjoint()

Checks whether two sets have no common elements.

issubset()

Checks whether all elements of one set are present in another set.

issuperset()

Checks whether a set contains all elements of another set.

Easy memory trick:

DISJOINT → No common
SUBSET   → Small set is inside big set
SUPERSET → Big set contains small set

### Frozenset
#### Frozen set is just an immutable version of a Python set object

```python
# create frozenset
fs1 = frozenset([1,2,3])
fs2 = frozenset([3,4,5])

fs1 | fs2
```

Important

Frozenset immutable hone ka matlab ye nahi ki operations nahi kar sakte.

Ye operations kar sakte ho:

fs1 | fs2    # Union ✅
fs1 & fs2    # Intersection ✅
fs1 - fs2    # Difference ✅
fs1 ^ fs2    # Symmetric difference ✅


set vs frozenset
Feature	        set	frozenset
Mutable	        ✅	❌
Add	            ✅	❌
Remove	        ✅	❌
Union	        ✅	✅
Intersection	✅	✅
Hashable	    ❌	✅
Dictionary key	❌	✅

A frozenset is an immutable version of a set. It cannot be modified after creation, but it supports set operations such as union, intersection, difference, and symmetric difference.

Most important point:
set → mutable → not hashable
frozenset → immutable → hashable

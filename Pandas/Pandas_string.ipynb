```python
# imort titanic
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df['Name']
```

```python
# common Functions 
# lower/uppper/capitalize/title
df['Name'].str.lower()
df['Name'].str.upper()
df['Name'].str.capitalize()
df['Name'].str.title()

# len
df['Name'][df['Name'].str.len()==82].values[0]

# strip
"            Nirbhay                                          ".strip()
df['Name'].str.strip()
```

```python
# split -> get
df['lastname']=df['Name'].str.split(',').str.get(0)
df.head()
```

```python
df[['title','firstname']]=df['Name'].str.split(',').str.get(1).str.strip().str.split(' ',n=1,expand=True)
df
```

```python
df['title'].value_counts()
```

```python
# Replace
df['title'].str.replace('Ms.','Miss.')
df['title'].str.replace('Mile.','Miss.')
```

```python
df['title'].value_counts()
```

```python
# Filtring
# startswith/endswith
df[df['firstname'].str.startswith('A')]
# isdigit/isalpha...
df[df['firstname'].str.isdigit()]
df[df['firstname'].str.isalpha()]

```

```python

# find lastnames with start and end char consonent 
df[df['firstname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```

```python
# applying regex
# contains
# search john -> both case
df[df['firstname'].str.contains('john',case=False)]
# find lastnames with start and end char vowel 
df[df['firstname'].str.contains('^[aeiouAEIOU].+[aeiouAEIOU]$')]
# ^ reprsent first charecter and $ represent last charecter. (.) dot reprsent any character + reprsent one or more 

```

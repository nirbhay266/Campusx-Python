### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
# string
country = ['India','Pakistan','USA','Nepal','Srilanka']

pd.Series(country)
```

```python
# Integers
marks = [90, 80, 70, 60, 50]
runs_ser=pd.Series(marks)
print(runs_ser)
```

```python
# Integers
marks = [90, 80, 70, 60, 50]
subjects = ['Maths', 'Science', 'English', 'History', 'Geography']
pd.Series(marks, index=subjects)
```

```python
# setting a name

marks = pd.Series(marks,index=subjects, name='Nirbhay yadav')
print(marks)
```

--The value_counts() method is used to count the frequency of each unique value in a Series. By default, results are sorted in descending order --of frequency.

#### Parameter	Use
    - normalize=True	Returns relative frequency (proportion) instead of counts.
    - sort=True	Sorts results by frequency (default: descending).
    - ascending=True	Sorts frequency in ascending order.
    - dropna=False	Includes NaN values in the count.

```python
# value_counts -> movies
movies.value_counts()
```

```python
print(movies.value_counts(normalize=True))
print(movies.value_counts(ascending=True))
print(movies.value_counts(sort=False))
print(movies.value_counts(dropna=False))
```

```python
# sort_values -> inplace
# sort_values(ascending=False) sorts values in descending order, head(1) selects the largest value, and .values[0] extracts that value as a scalar.
 
vk.sort_values(ascending=False).head(1).values[0]
```

```python
# inplace=False → Original data unchanged (default in sort_values).
# inplace=True → Original data directly modify.
vk.sort_values(ascending=False, inplace=True)
```

## Filtering a DataFrame

```python
ipl.head(2)
```

```python
# find all the final winners
mask = ipl['MatchNumber'] == 'Final'
new_df = ipl[mask]
new_df[['Season','WinningTeam']]

ipl[ipl['MatchNumber'] == 'Final'][['Season','WinningTeam']] 
```

```python
# how many super over finishes have occured 
ipl[ipl['SuperOver'] == 'Y'].shape[0]


# .shape returns a tuple containing (rows, columns).
# .shape[0] returns the total number of rows.
# .shape[1] returns the total number of columns.
```

```python
# How many matches has csk won in kolkata
ipl[(ipl['City'] == 'Kolkata') & (ipl['WinningTeam'] == 'Chennai Super Kings')].shape[0]
```

```python
# toss winner is match winner in percentage 
ipl[ipl['TossWinner'] == ipl['WinningTeam']].shape[0] / ipl.shape[0] * 100
```

```python
# movies with rating higher than 8 and votes >10000

movies[(movies['imdb_rating'] > 8) & (movies['imdb_votes'] > 10000)].shape[0]
```

##### str.contains() checks whether a specific word or text is present in each string of a Pandas Series.

```python
# Action works with rating higher than 7.5
# mask1 = movies['genres'].str.split('|').apply(lambda x: 'Action' in x)

mask1 = movies['genres'].str.contains('Action')
mask2 = movies['imdb_rating'] > 7.5

movies[mask1 & mask2]
```

```python
ipl['ID'] = ipl['ID'].astype('int32')
```

```python
ipl.info()
```

Category Data Type: A Pandas data type used to store data with a limited number of unique values. It maintains unique categories and represents each value using an integer code. It can reduce memory usage when values are repeated frequently.

```python
# ipl['Season'] = ipl['Season'].astype('category')
ipl['Team1'] = ipl['Team1'].astype('category')
ipl['Team2'] = ipl['Team2'].astype('category')
```

```python
ipl.info()
```

```python
# value_counts
ipl['Team2'].value_counts()
```

```python
# find which player has won most potm -> in finals and qualifiers
ipl[ipl['MatchNumber'].str.contains(
    'Final|Qualifier', na=False
)]['Player_of_Match'].value_counts().head(10)
```

```python
# Toss decision plot
ipl['TossDecision'].value_counts().plot(kind='bar', title='Toss Decision')
```

```python
# how many matches each team has played
team_matches =ipl['Team1'].value_counts() + ipl['Team2'].value_counts()
team_matches  = team_matches.sort_values(ascending=False)
team_matches
```

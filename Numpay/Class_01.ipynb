## What is numpay?

NumPy is the fundamental package for scientific computing in python. It is a Python library that provides a multidimentional array object, various derived object (such as masked array and metrices), and a assortment of routines for fast operations on arrays, including mathematical, logical, shape manipulation, sorting, selecting, I/O, descrete Fourier transforms, basic linear algebra, basic statistical operations, random simulation and much more.  

### Numpy Arrays Vs Python Sequences

* NumPy arrays have a fixed size at creation, unlike Python lists(which can grow dynamically). Changing the size of an ndarray will create a new array and delete the original.

* The element in a NumPy array are all requaired to be of the same data type, and this will be the same size in memory.

* NumPy arrays facilitate advanced mathematical and other types of operations on large numbers of data. Typically, such operations are executed more efficiently and with less code than is possible using Python's built- in sequences.

* A growing plethora of scientific and mathematical Python-based packages are using NumPy arrays; though these typicaly support Python-sequence input, they convery such input to NumP arrays prior to processing, and they often output NumPy arrays.

### Array Functions

```python
import numpy as np
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# min/max/sum/prod
# 0->col and 1 ->row
np.max(a1)
np.max(a1,axis=0)
```

```python
# mean/median/std/var
np.mean(a1,axis=1)
np.median(a1,axis=0)
np.std(a1,axis=0)
np.var(a1,axis=0)
```

```python
# dot product
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)

np.dot(a2,a3)

```

```python
# log and exponents
np.log(a1)
```

```python
np.exp(a1)
```

```python
# round/floor/ceil
np.random.random((2,3))*100
np.round(np.random.random((2,3))*100)
```

```python
# round/floor/ceil
np.random.random((2,3))*100
np.floor(np.random.random((2,3))*100)
```

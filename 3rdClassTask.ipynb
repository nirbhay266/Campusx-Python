Problem 1 - Print the following pattern. Write a program to use for loop to print the following reverse number pattern.
5 4 3 2 1 
4 3 2 1 
3 2 1 
2 1 
1

```python
for i in range(0,5):
    for j in range(5-i,0,-1):
        print(j, end="")
    print()
     
```

Problem 2: Print the following pattern.
* 
* * 
* * * 
* * * * 
* * * * * 
* * * * 
* * * 
* * 
*

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()

for i in range(4, 0, -1):
    for j in range(i):
        print("*", end=" ")
    print()
```

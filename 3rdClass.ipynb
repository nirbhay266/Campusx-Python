##### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years.

```python
# Code here
curr_pop = 10000

for i in range(10,0,-1):
  print(i,curr_pop)
  curr_pop = curr_pop/1.1
```

#### Sequence sum
1/1! + 2/2! + 3/3! + ...

```python
n = int(input('enter n'))

result = 0
fact = 1

for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact

print(result)
```

### Loop Control Statement
1. Break
2. Continue
3. Pass


```python
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))

for i in range(lower,upper+1):
  for j in range(2,i):
    if i%j == 0:
      break
  else:
    print(i)
```

```python
# Continue
for i in range(1,10):
  if i == 5:
    continue
  print(i)
```

```python
for i in range(1,10):
  pass
  
```

## Creating Stings

```python
s = 'hello'
s = "hello"
# multiline strings
s = '''hello'''
s = """hello"""
s = str('hello')
print(s)
```

```python
# Positive Indexing
s = 'hello world'
print(s[4])
```

```python
# negative Indexing
s = 'hello world'
print(s[-1])
print(s[-2])
print(s[6:0:-2])
print(s[::-1])
print(s[0:6:2])
print(s[1::])
```

```python
# Editing and Deleting in Strings
s='Hello World' \
s[1]='P'
print(s)
# python string are immutable, we cannot change the value of string once it is created. We can create a new string with the desired changes.
```

```python
s = 'hello world'
del s
print(s)
```

```python
s = 'hello world'
del s[-1:-5:2]
print(s)
```

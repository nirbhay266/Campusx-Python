### Namespaces

A namespace is a space that holds names(identifiers). Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values).

* There are 4 types of namespaces:
    - Builtin Namespace
    - Global Namespace
    - Enclosing Namespace
    - Local Namespace

### Scope and LEGB Rule

- A scope is a textual region of a Python program where a namespace is directly accessible.
- The iterpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scop. 
- If the intyerpreter doesn't find the name in any of these locations, then Python raise a NameErro exception.

```python
# local and global
# global var
a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> same name
a = 2

def temp():
  # local var
  a = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> local does not have but global has
a = 2

def temp():
  # local var
  print(a)

temp()
print(a)
```

```python
# local and global -> editing global
a = 2

def temp():
  # local var
  a += 1
  print(a)

temp()
print(a)
```

```python
a = 2

def temp():
  # local var
  global a
  a += 1
  print(a)

temp()
print(a)
```

```python
# local and global -> global created inside local
def temp():
  # local var
  global a
  a = 1
  print(a)

temp()
print(a)
```

```python
# local and global -> function parameter is local
def temp(z):
  # local var
  print(z)

a = 5
temp(5)
print(a)
print(z)
```

## Decorators
- A decorator in python is a function that receive another function as input and adds some functionality(decoration) to and it and reuturn it.
- This can happen only because python functions are 1st class citizens.
- There are 2 types of decorators avilable in python.
    - Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property atc.
    - User defined decorators that we programmers can create according to our needs.

```python
# Python are 1st class function
def modify1(func,num):
    return func(num)

def square(num):
    return num**2

modify1(square,2)
```

```python
# simple example
def my_decorator(func):
    def wrapper():
        print('************************************')
        func()
        print('************************************')
    return wrapper

def hello():
    print('Hello')
def display():
    print('Hello Nirbhay : ')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
# Better syntax?
# simple example

def my_decorator(func):
    def wrapper():
        print('************************************')
        func()
        print('************************************')
    return wrapper

@my_decorator
def hello():
    print('Hello')

hello()
```

```python
# anything meaningful?
import time

def timer(func):
  def wrapper(*args):
    start = time.time()
    func(*args)
    print('time taken by',func.__name__,time.time()-start,'secs')
  return wrapper

@timer
def hello():
  print('hello wolrd')
  time.sleep(2)

@timer
def square(num):
  time.sleep(1)
  print(num**2)

@timer
def power(a,b):
  print(a**b)

hello()
square(2)
power(2,3)
```

```python
# A big problem
def sanity_check(data_type):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if type(args[0]) == data_type:
                func(*args)
            else:
                raise TypeError('Ye datatype nhi chaleg : ')
        return inner_wrapper
    return outer_wrapper

@sanity_check(int)
def square(num):
    print(num**2)

@sanity_check(str)
def geet(name):
    print("Hello ",name)

geet('nirbhay')
```

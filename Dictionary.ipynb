# Dictionary

    Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

    In some languages it is known as map or assosiative arrays.

    dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

    Characterstics:

    - Mutable
    - Indexing has no meaning
    - keys can't be duplicated
    - keys can't be mutable items

Dictionary
│
├── Key-Value pairs
├── Mutable
├── Keys unique
├── Keys hashable
├── Values duplicate ho sakti hain
├── Key-based access
├── Python 3.7+ → insertion order maintained
└── Fast average lookup by key

### Create Dictionary

```python
# empty dictionary
d={}
print(d)

# 1D dictionary
d1={'name':'nirbhay','gender':'male'}
print(d1)

# with mixed keys
d2={'name':'nirbhay','gender':'male',1:[1,2,3,4],'age':20}
print(d2)

# 2D dictionary -> JSON
s={
    'name':'nirbhay',
    'gender':'male',
    'college':'Ms',
        'subject':{
                'dsa':50,
                'math':67,
                'english':90
        }

}
print(s)
print(s['subject']['math']) # accessing 2D dictionary
print(s['college']) # accessing 1D dictionary

# using sequence and dict functions
d4=dict([('name','nirbhay'),('age',21),(3,3)])
print(d4)

# Duplicate keys
d5={'name':'nirbhay','gender':'male','name':'kumar'}
print(d5) # last value of duplicate key will be considered

# mutable items as keys
d6 = {'name':'nitish',(1,2,3):2}
print(d6)
```

### Accessing items

```python
my_dict = {'name': 'Jack', 'age': 26}
# []
my_dict['age']
# get
my_dict.get('age')

s['subject']['math']
```

### Adding key-value pair

```python
d4['gender'] = 'male'
d4
d4['weight'] = 72
print(d4)

s['subject']['dsa'] = 75
s
```

### Remove key-value pair

```python
d = {'name': 'nitish', 'age': 32, 3: 3, 'gender': 'male', 'weight': 72}
print(d)
# pop 
print(d.pop('age')) # remove and return the value of the key
print(d)
# popitem
print(d.popitem()) # remove and return the last key-value pair
print(d)
del d['name'] # delete a key-value pair
print(d)
# clear
d.clear() # remove all key-value pairs
print(d)
```

#### Dictionary Operations

        - Iteration

```python
print(s)
'name' in s
```

```python
d = {'name':'nitish','gender':'male','age':33}

for i in d:
  print(i,d[i]) # print key and value
```

#### Dictionary Functions

```python
# len/sorted
print(len(d)) # number of key-value pairs
print(sorted(d)) # print keys in sorted order
print(sorted(d.values())) # print values in sorted order # Error: Python int aur str ko directly compare karke sort nahi kar sakta.
```

```python
d = {
    "math": 80,
    "python": 95,
    "sql": 85
}

print(len(d))
print(sorted(d))
print(sorted(d.values()))
```

```python
# Update 
d1={1:2,3:4,4:5}
d2={4:7,6:8}

d1.update(d2) # update d1 with d2
print(d1)
```

#### Dictionary Comprehension
{key:value for vars in iterable}

```python
# print 1st 10 number and their square in dictionary
{i:i**2 for i in range(1,11)}
```

```python
distances={'delhi':100,'mumbai':2000,'banglore':1500}
print(distances.items())
```

```python
# using existing dictionary to create new dictionary
distance={'delhi':100,'mumbai':2000,'banglore':1500}
{key:value*0.62 for(key,value) in distance.items()}
```

```python
# using zip
days=["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday"]
temps=[25,28,30,27,26,29]
{i:j for (i,j) in zip(days,temps)}
```

```python
# Using If condition in dictionary comprehension
product={'phone':10,'laptop':5,'tablet':0}
{key:value for key,value in product.items() if value > 0}
```

```python
# Nested Comprehension 
# print tables of number from 2 to 5
{i:{j:i*j for j in range(1,11)} for i in range(2,6)}
```

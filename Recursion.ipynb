```python
def mul(a,b):
    if b == 1:
        return a
    else:
        return a + mul(a,b-1)

print(mul(3,4))
```

```python
def fact(number):
    if number == 1:
        return number
    else:
        return number*fact(number-1)

fact(5)
```

```python
# Check palindrom 
def palin(text):
    if len(text) <= 1:
        print("Palindrome ")
    else:
        if text[0] == text[-1]:
            palin(text[1:-1])
        else:
            print("Not a Palindrome : ")
palin("madam")
palin('abba')
```

```python
def fib(m):
    if m==0 or m==1:
        return 1
    else:
        return fib(m-1) + fib(m-2)
print(fib(5))
```

```python
def memo(m,d):
    if m in d:
        return d[m]
    else:
        d[m] = memo(m-1,d) + memo(m-2,d)
        return d[m]

d = {0:1,1:1}
print(memo(48,d))
```

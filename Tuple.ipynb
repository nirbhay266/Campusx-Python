### Creating Tuples

```python
# emplty
t1=()
print(t1)
# create a tuple with a single item
t2=('hello')
print(t2)
print(type(t2))

# homo
t3=(1,2,3,4,5)
# Hetro
t4=(1,2,5,True,[1,2,3,4])
print(t4)
# tuple
t5=(1,2,3,(4,5))
print(t5)
# using type converstion
t6=tuple('hello')
print(t6)
```

## Difference between Lists and Tuples
##### Syntax
1. List → []
2. Tuple → ()
##### Mutability
1. List → Mutable, yani elements ko change/add/remove kar sakte hain.
2. Tuple → Immutable, yani create hone ke baad change nahi kar sakte.
##### Speed
1. List → Comparatively slower.
2. Tuple → Comparatively faster.
##### Memory
1. List → More memory consume karti hai.
2. Tuple → Less memory consume karta hai.
##### Built-in functionality
1. List → More methods available, jaise append(), extend(), insert(), remove(), pop(), sort().
2. Tuple → Fewer methods, mainly count() aur index().
##### Error prone
1. List → Mutable hone ki wajah se accidentally modify ho sakti hai.
2. Tuple → Immutable hone ki wajah se accidental modification ka risk kam hota hai.
##### Usability
1. List → Jab data ko modify karna ho.
2. Tuple → Jab data fixed/constant rakhna ho.

```python
## Speed
import time 

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
```

```python
## Memory
import sys 
L=list(range(1000))
T=tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

#### Why use tuple?
We use tuples when we need an ordered collection of values that should not be modified. Tuples are immutable, can be used as dictionary keys when their elements are hashable, and generally have lower memory overhead than lists.

```python
# tuple unpacking
a,b,c = (1,2,3)
print(a,b,c)
```

```python
a,b = (1,2,3)
print(a,b)
```

```python
a = 1
b = 2
a,b = b,a

print(a,b)
```

```python
a,b,*others=(1,2,3,4)
print(a,b)
print(others)
```

```python
# zipping tuple 
a=(1,2,3,4,5)
b=(6,7,8,9,10)
tuple(zip(a,b))
```
